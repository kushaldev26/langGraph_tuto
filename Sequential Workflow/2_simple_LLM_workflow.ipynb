```python
from langgraph.graph import StateGraph,START,END
from langchain_mistralai import ChatMistralAI
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
from typing import TypedDict
```

```python
load_dotenv()
```

```python
# LLM - model 
llm = ChatMistralAI(model='mistral-large-latest')
```

```python
# create state
class LLMState:

    question: str
    answer: str
```

```python
def llm_qa(state: LLMState)->LLMState:

    # extraxt the question from state
    query = state['question']

    # from a prompt
    prompt = f"Aswer the following Question {query}"

    # ask that questio to LLM
    responce = llm.invoke(prompt).content

    # update that answer to the state
    state['answer'] = responce

    return state
```

```python
# define the graph
graph = StateGraph(LLMState)
```

```python
# add node in the graph
# graph.add_node(START)
graph.add_node('llm_qa',llm_qa)
```

```python
# add edges in the graph
graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)
```

```python
# compile the graph
workflow = graph.compile()
```

```python
initial_state = {"question":"How far moon from the earth"}

final_state = workflow.invoke(initial_state)
```
